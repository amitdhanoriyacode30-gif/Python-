```python
print("------SIMPLE CALCULATER-----")

# TAKE first username num1
num1 = float(input("Enter the first value:"))
operator = input("Enter operator(+,-,*,/):")
num2 = float(input("Enter the first value: "))

#Use the  condition  for check the posssiblily

if operator == "+":
    result = num1 + num2

elif operator == "-":
    result = num1 - num2
elif operator == "*":
    result = num1 * num2

elif operator == "/":
    if num2 != 0:
        result = num1 / num2
    else:
        result = "Error! Cannot divide by zero."
else:
    result = "Invalid operator."


print("Result:",result)

```
